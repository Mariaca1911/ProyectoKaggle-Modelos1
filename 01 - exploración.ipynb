```python
import pandas as pd
```

```python
dtr = pd.read_csv('train.csv')
dtr
```

```python
dtr.E_VALORMATRICULAUNIVERSIDAD.value_counts()
```

```python
import matplotlib.pyplot as plt

# Gráfico de barras de las horas trabajadas por semana
dtr.E_HORASSEMANATRABAJA.value_counts().plot.bar(color='skyblue', edgecolor='black')

plt.title('Distribución de horas trabajadas por semana')
plt.xlabel('Rango de horas semanales')
plt.ylabel('Número de estudiantes')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Gráfico de barras del estrato de vivienda
dtr.F_ESTRATOVIVIENDA.value_counts().plot.bar(color='lightgreen', edgecolor='black')

plt.title('Distribución por estrato de vivienda')
plt.xlabel('Estrato socioeconómico')
plt.ylabel('Número de estudiantes')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Contar los programas por departamento
departamentos = dtr.E_PRGM_DEPARTAMENTO.value_counts().head(10)  # muestra solo los 10 primeros para que sea más legible

# Crear la gráfica
departamentos.plot(kind='bar', color='lightcoral', edgecolor='black')

plt.title('Top 10 Departamentos con más programas académicos')
plt.xlabel('Departamento')
plt.ylabel('Cantidad de estudiantes')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
